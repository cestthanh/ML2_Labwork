```python
import matplotlib.pyplot as plt

# Giả lập dữ liệu (thay bằng dữ liệu thật của bạn)
k_values = list(range(1, 21))
accuracies = [0.905, 0.88, 0.88, 0.905, 0.93, 0.88, 0.905, 0.905, 0.905, 0.905,
              0.905, 0.88, 0.88, 0.88, 0.855, 0.88, 0.905, 0.905, 0.88, 0.93]

# Tạo biểu đồ
plt.figure(figsize=(4.5, 3))  # thu nhỏ lại

plt.plot(k_values, accuracies, marker='o', markersize=4, linewidth=1)

# Chỉnh title và label nhỏ hơn, đơn giản hơn
plt.title("Accuracy vs. Number of Neighbors (k)", fontsize=10)
plt.xlabel("Number of Neighbors (k)", fontsize=9)
plt.ylabel("Accuracy", fontsize=9)

# Chỉnh các ticks nhỏ và đơn giản
plt.xticks(fontsize=8)
plt.yticks(fontsize=8)

# Tắt grid hoặc để grid nhẹ
plt.grid(True, linestyle='--', linewidth=0.3, alpha=0.5)

# Không dùng tight_layout
plt.show()
```
