```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder, OneHotEncoder
```

DATASET SUPERSTORE DATA

```python
df = pd.read_csv("superstore_cleaned.csv")
df.head()
```

```python
df.info()
df.describe()
```

```python
# Sao chép dataframe gốc
df_encoded = df.copy()


# Định nghĩa thứ tự cho Education
education_order = ['Basic', '2n Cycle', 'Graduation', 'Master', 'PhD']
# Khởi tạo OrdinalEncoder với thứ tự category
ordinal_encoder = OrdinalEncoder(categories=[education_order])
# Áp dụng Ordinal Encoding cho Education và chuyển kiểu sang int
df_encoded['Education'] = ordinal_encoder.fit_transform(df_encoded[['Education']]).astype(int)


# Initialize encoders
onehot_encoder = OneHotEncoder(sparse_output=False)

# Áp dụng One-Hot Encoding cho Marital_Status trên dataframe đã sao chép
marital_encoded = onehot_encoder.fit_transform(df_encoded[['Marital_Status']])
# Tạo DataFrame chứa các cột one-hot encode, đặt tên cột tự động theo Marital_Status
marital_encoded_df = pd.DataFrame(marital_encoded, columns=onehot_encoder.get_feature_names_out(['Marital_Status']))
# Ghép DataFrame one-hot vào dataframe gốc đã loại bỏ cột Marital_Status
df_encoded = pd.concat([df_encoded.drop(columns=['Marital_Status']), marital_encoded_df], axis=1)

# Kiểm tra kết quả
print(df_encoded.head())
print(df_encoded.columns)



```

```python
df_encoded.to_csv("superstore_encoded.csv", index = False)
```
