```python
import pandas as pd

# Đọc dữ liệu từ file CSV gốc
df = pd.read_csv("diabetes.csv")  # Thay đường dẫn nếu cần

# Danh sách các cột không nên có giá trị bằng 0
cols_with_invalid_zeros = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']

# Thay thế các giá trị 0 bằng NaN
df[cols_with_invalid_zeros] = df[cols_with_invalid_zeros].replace(0, pd.NA)

# Điền giá trị thiếu bằng trung vị của từng cột
df[cols_with_invalid_zeros] = df[cols_with_invalid_zeros].fillna(df[cols_with_invalid_zeros].median())

# Xuất ra file mới
df.to_csv("diabetes_cleaned.csv", index=False)

print("✅ Đã xử lý và lưu dữ liệu vào 'diabetes_cleaned.csv'")
```
